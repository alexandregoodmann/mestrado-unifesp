# **hist_RGB.py**

```python
import cv2 #importa biblioteca OpenCV

import numpy as np
from matplotlib import pyplot as grafico

#Carregando e mostrando imagem
imagem = cv2.imread("folha.jpg")
cv2_imshow(imagem)

#Divide imagem em canais RGB
azul, verde, vermelho = cv2.split(imagem)
cv2_imshow( vermelho)
cv2_imshow(verde)
cv2_imshow(azul)

#Histogramas com 256 classes
grafico.title("Azul")
grafico.hist(azul.ravel(), 256, [0,256], color='b')
grafico.figure();

grafico.title("Verde")
grafico.hist(verde.ravel(), 256, [0,256], color='g')
grafico.figure();

grafico.title("Vermelho")
grafico.hist(vermelho.ravel(), 256, [0,256], color='r')
grafico.show();
```

# **equalizacao_Hist.py**

```python
import cv2 #importa biblioteca OpenCV

import numpy as np
from matplotlib import pyplot as grafico

#Carregando e mostrando imagem
imagemOriginal = cv2.imread("nevoa.jpg", 0)
cv2_imshow( imagemOriginal)

#Gera Histograma
grafico.title("Histograma Original")
grafico.hist(imagemOriginal.ravel(), 100, [0,256], color = 'k')
grafico.show()

#Gera imagem com histograma Equalizado
imagemEqua = cv2.equalizeHist(imagemOriginal)
print("Imagem Equalizada")
cv2_imshow(imagemEqua)

#Gera Histograma da nova imagem
grafico.title("Histograma Equalizado")
grafico.hist(imagemEqua.ravel(), 256, [0,256], color = 'k')
grafico.show()

cv2.imwrite("nevoaEqua.jpg", imagemEqua)
```

# **CLAHE.PY**

```python
import cv2 #importa biblioteca OpenCV

import numpy as np
from matplotlib import pyplot as grafico

#Carregando e mostrando imagem
imagem = cv2.imread("nevoa.jpg", 0)
cv2_imshow(imagem)
#Histograma com da imagem original
grafico.title("Original")
grafico.hist(imagem.ravel(), 256, [0,256], color = "k")
grafico.show()

#Gera imagem com histograma Equalizado (global)
imagemEqua = cv2.equalizeHist(imagem)
print("Imagem Equalizada")
cv2_imshow( imagemEqua)
#Gera seu Histograma
grafico.title("Histograma Equalizado")
grafico.hist(imagemEqua.ravel(), 256, [0,256], color = 'k')
grafico.show()

# Cria um objeto CLAHE
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
imagemClahe = clahe.apply(imagem)
cv2.imwrite("Clahe.jpg", imagemClahe)
print("Após CLAHE")
cv2_imshow(imagemClahe)
#Mostra histograma da imagem após CLAHE
grafico.title("Após Clahe")
grafico.hist(imagemClahe.ravel(), 256, [0,256], color="k")
grafico.show()


```

# **CriaImagemGradiente.py**

```python
import numpy as np
import cv2 as cv

# Define o tamanho da imagem que será criada
n = 100  # Número de linhas
m = 100  # Número de colunas

# Defini o incremento para gerar o gradiente
r = 255/(m-1)

# Cria a image (matrix) do tamanho desenjado (escala de cinza)
img = np.empty((n,m),np.uint8)

# Percorre os pixels da matriz atribuindo o valor calculado
for i in range(0, n):
    for j in range(0,m):
        img[i][j] = j*r

# Escreve a imagem
cv.imwrite("gradiente.png", img)  # Guarda la imagen
cv2_imshow(img)                   # Muestra la matriz de la imagen
```

# **LimiarizacaoSimples.py**

```python
import cv2 as cv
from matplotlib import pyplot as plt
# Carrega uma imagem (escala de cinza)
img = cv.imread('gradiente.png',0)
lim = 50

# Aplica diversos tipos de limiarização (Threshold)
ret,thresh1 = cv.threshold(img,lim,255,cv.THRESH_BINARY)
ret,thresh2 = cv.threshold(img,lim,255,cv.THRESH_BINARY_INV)
ret,thresh3 = cv.threshold(img,lim,255,cv.THRESH_TRUNC)
ret,thresh4 = cv.threshold(img,lim,255,cv.THRESH_TOZERO)
ret,thresh5 = cv.threshold(img,lim,255,cv.THRESH_TOZERO_INV)
# Mostra o resultado
titles = ['Imagem Original','Binária','Binário Invertido',\
          'Trubcado','Para Zero','Para Zero Invertido']
images = [img, thresh1, thresh2, thresh3, thresh4, thresh5]
# Mostra as imagens
for i in range(6):
   plt.subplot(2,3,i+1)
   plt.imshow(images[i],"gray")
   plt.title(titles[i])
   plt.xticks([])
   plt.yticks([])
plt.show()

# Necessário instalar python3-tk para rodar esse script usando o comando
# sudo apt install python3-tk
```

# **LimiarizacaoEmFaixaBarra.py**

```python
import cv2 as cv
from google.colab.patches import cv2_imshow
import ipywidgets as widgets
from ipywidgets import interact

# --------------------------
# 1. Cargar la imagen
# --------------------------
# Asegúrate de que el archivo c3.jpg esté subido al entorno de Colab
frame = cv.imread("c3.jpg")

if frame is None:
    raise FileNotFoundError("No se encontró 'c3.jpg'. Sube la imagen con files.upload()")

# Convertir a HSV
frame_HSV = cv.cvtColor(frame, cv.COLOR_BGR2HSV)

# --------------------------
# 2. Función para aplicar segmentación
# --------------------------
def segmentar(menor_H, maior_H, menor_S, maior_S, menor_V, maior_V):
    # Aplicar el filtro por rango
    frame_threshold = cv.inRange(frame_HSV,
                                 (menor_H, menor_S, menor_V),
                                 (maior_H, maior_S, maior_V))

    # Mostrar resultados
    print("Imagen segmentada (threshold):")
    cv2_imshow(frame_threshold)

# --------------------------
# 3. Sliders interactivos
# --------------------------
interact(segmentar,
         menor_H=widgets.IntSlider(min=0, max=179, step=1, value=0, description="Menor H"),
         maior_H=widgets.IntSlider(min=0, max=179, step=1, value=179, description="Maior H"),
         menor_S=widgets.IntSlider(min=0, max=255, step=1, value=0, description="Menor S"),
         maior_S=widgets.IntSlider(min=0, max=255, step=1, value=255, description="Maior S"),
         menor_V=widgets.IntSlider(min=0, max=255, step=1, value=0, description="Menor V"),
         maior_V=widgets.IntSlider(min=0, max=255, step=1, value=255, description="Maior V"))
```
